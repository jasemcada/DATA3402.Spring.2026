1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2 and isinstance(args[0], int) and isinstance(args[1], int):
            n, m = args
            self._matrix = [[0.0 for _ in range(m)] for _ in range(n)]
        elif len(args) == 1 and isinstance(args[0], list):
            data = args[0]
            if not self._is_valid_matrix(data):
                raise ValueError("All rows must have the same number of columns")
            self._matrix = [[float(val) for val in row] for row in data]
        else:
            raise ValueError("Invalid arguments for matrix initialization")
    
    def _is_valid_matrix(self, data):
        if not data or not isinstance(data[0], list):
            return False
        row_length = len(data[0])
        for row in data:
            if len(row) != row_length:
                return False
        return True
    
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self._matrix[i][j]
        else:
            return self._matrix[key]
    
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._matrix[i][j] = value
        else:
            self._matrix[key] = value
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    return matrix([[c for _ in range(m)] for _ in range(n)])

def zeros(n, m):
    return matrix(n, m)

def ones(n, m):
    return matrix([[1.0 for _ in range(m)] for _ in range(n)])

def eye(n):
    result = [[0.0 for _ in range(n)] for _ in range(n)]
    for i in range(n):
        result[i][i] = 1.0
    return matrix(result)
```

```python
C = constant(2, 3, 5.0)
print("Constant (2x3, value=5):")
print(C.to_list())

Z = zeros(2, 3)
print("\nZeros (2x3):")
print(Z.to_list())

O = ones(2, 3)
print("\nOnes (2x3):")
print(O.to_list())

I = eye(3)
print("\nIdentity (3x3):")
print(I.to_list())
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self._matrix = [[0.0 for _ in range(m)] for _ in range(n)]
        else:
            data = args[0]
            row_length = len(data[0])
            for row in data:
                if len(row) != row_length:
                    raise ValueError("All rows must have same length")
            self._matrix = [[float(val) for val in row] for row in data]
    
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self._matrix[i][j]
        else:
            return self._matrix[key]
    
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._matrix[i][j] = value
        else:
            self._matrix[key] = value
    
    def shape(self):
        n = len(self._matrix)
        m = len(self._matrix[0])
        return (n, m)
    
    def transpose(self):
        n, m = self.shape()
        result = [[self._matrix[j][i] for j in range(n)] for i in range(m)]
        return matrix(result)
    
    def row(self, n):
        return matrix([self._matrix[n]])
    
    def column(self, n):
        result = [[self._matrix[i][n]] for i in range(len(self._matrix))]
        return matrix(result)
    
    def to_list(self):
        return self._matrix
    
    def block(self, n_0, n_1, m_0, m_1):
        result = []
        for i in range(n_0, n_1):
            row = []
            for j in range(m_0, m_1):
                row.append(self._matrix[i][j])
            result.append(row)
        return matrix(result)

    def scalarmul(self, c):
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(c * self[i, j])
            result.append(row)
        return matrix(result)

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] + N[i, j])
            result.append(row)
        return matrix(result)

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] - N[i, j])
            result.append(row)
        return matrix(result)

    def mat_mult(self, N):
        n1, m1 = self.shape()
        n2, m2 = N.shape()

        if m1 != n2:
            raise ValueError("Inner dimensions must agree")

        result = []
        for i in range(n1):
            row = []
            for j in range(m2):
                total = 0.0
                for k in range(m1):
                    total += self[i, k] * N[k, j]
                row.append(total)
            result.append(row)
        return matrix(result)

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] * N[i, j])
            result.append(row)
        return matrix(result)

    def equals(self, N):
        if self.shape() != N.shape():
            return False

        n, m = self.shape()
        for i in range(n):
            for j in range(m):
                if self[i, j] != N[i, j]:
                    return False
        return True
```

```python
A = matrix([[1,2],[3,4]])
B = matrix([[5,6],[7,8]])
C = matrix([[2,0],[1,2]])

print("A =", A.to_list())
print("B =", B.to_list())
print("C =", C.to_list())

print("\nScalar Multiplication (2A):")
print(A.scalarmul(2).to_list())

print("\nAddition (A + B):")
print(A.add(B).to_list())

print("\nSubtraction (B - A):")
print(B.sub(A).to_list())

print("\nMatrix Multiplication (A * C):")
print(A.mat_mult(C).to_list())

print("\nElement-wise Multiplication (A .* B):")
print(A.element_mult(B).to_list())

print("\nEquality Test (A == A):")
print(A.equals(A))

print("\nEquality Test (A == B):")
print(A.equals(B))
```

```python
class matrix:
    def __init__(self, *args):
        if len(args) == 2:
            n, m = args
            self._matrix = [[0.0 for _ in range(m)] for _ in range(n)]
        else:
            data = args[0]
            row_length = len(data[0])
            for row in data:
                if len(row) != row_length:
                    raise ValueError("All rows must have same length")
            self._matrix = [[float(val) for val in row] for row in data]
    
    def __getitem__(self, key):
        if isinstance(key, tuple):
            i, j = key
            return self._matrix[i][j]
        else:
            return self._matrix[key]
    
    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._matrix[i][j] = value
        else:
            self._matrix[key] = value
    
    def shape(self):
        n = len(self._matrix)
        m = len(self._matrix[0])
        return (n, m)
    
    def transpose(self):
        n, m = self.shape()
        result = [[self._matrix[j][i] for j in range(n)] for i in range(m)]
        return matrix(result)
    
    def row(self, n):
        return matrix([self._matrix[n]])
    
    def column(self, n):
        result = [[self._matrix[i][n]] for i in range(len(self._matrix))]
        return matrix(result)
    
    def to_list(self):
        return self._matrix
    
    def block(self, n_0, n_1, m_0, m_1):
        result = []
        for i in range(n_0, n_1):
            row = []
            for j in range(m_0, m_1):
                row.append(self._matrix[i][j])
            result.append(row)
        return matrix(result)

    def scalarmul(self, c):
        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(c * self[i, j])
            result.append(row)
        return matrix(result)

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] + N[i, j])
            result.append(row)
        return matrix(result)

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] - N[i, j])
            result.append(row)
        return matrix(result)

    def mat_mult(self, N):
        n1, m1 = self.shape()
        n2, m2 = N.shape()

        if m1 != n2:
            raise ValueError("Inner dimensions must agree")

        result = []
        for i in range(n1):
            row = []
            for j in range(m2):
                total = 0.0
                for k in range(m1):
                    total += self[i, k] * N[k, j]
                row.append(total)
            result.append(row)
        return matrix(result)

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Matrix sizes must match")

        n, m = self.shape()
        result = []
        for i in range(n):
            row = []
            for j in range(m):
                row.append(self[i, j] * N[i, j])
            result.append(row)
        return matrix(result)

    def equals(self, N):
        if self.shape() != N.shape():
            return False

        n, m = self.shape()
        for i in range(n):
            for j in range(m):
                if self[i, j] != N[i, j]:
                    return False
        return True

    def __add__(self, N):
        return self.add(N)

    def __sub__(self, N):
        return self.sub(N)

    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)
        else:
            return self.scalarmul(other)

    def __rmul__(self, other):
        return self.scalarmul(other)

    def __eq__(self, N):
        return self.equals(N)
```
